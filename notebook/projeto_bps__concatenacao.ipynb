ANÁLISE EXPLORATÓRIA DE DADOS - PROJETO 'BPS'
_______________________________________________________________________________________________________________________

O objetivo dessa AED é concatenar e tratar as informações de toda a base de dados do Banco de Preços em Saúde - BPS criado pelo Ministério da Saúde para, ao final, embasar um dashboard contendo as principais informações acerca das aquisições de medicamentos e produtos médicos pelo setor público entre os anos de 2020 e 2026.


CONCATENANDO A BASE DE DADOS
_______________________________________________________________________________________________________________________

```python
# importando as bibliotecas necessárias para a concatenação dos dataframes
import pandas as pd

# importando a base de dados diretamente do GitHub - repositório "mini_projeto_bps"
URL_query1 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2020.csv?raw=true'
URL_query2 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2021.csv?raw=true'
URL_query3 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2022.csv?raw=true'
URL_query4 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2023.csv?raw=true'
URL_query5 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2024.csv?raw=true'
URL_query6 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2025.csv?raw=true'
URL_query7 = 'https://github.com/mariahbarros/mini_projeto_bps/blob/main/dados/dados_bps_2026.csv?raw=true'

# transformando as bases de dados em dataframes
df_query1 = pd.read_csv(URL_query1, sep=';', encoding='utf-8')
df_query2 = pd.read_csv(URL_query2, sep=';', encoding='utf-8')
df_query3 = pd.read_csv(URL_query3, sep=';', encoding='utf-8')
df_query4 = pd.read_csv(URL_query4, sep=';', encoding='utf-8')
df_query5 = pd.read_csv(URL_query5, sep=';', encoding='utf-8')
df_query6 = pd.read_csv(URL_query6, sep=';', encoding='utf-8')
df_query7 = pd.read_csv(URL_query7, sep=';', encoding='utf-8')


# formatação para exibir os dataframes de forma mais legível
pd.set_option('display.width', None)          
pd.set_option('display.max_columns', None)    
pd.set_option('display.expand_frame_repr', False)  

# reunindo todos os dataframes em uma lista
dados_bps_20_26 = [
    df_query1, 
    df_query2, 
    df_query3, 
    df_query4, 
    df_query5, 
    df_query6, 
    df_query7
]

# concatenar todos os dataframes em um único dataframe
tabela_unificada = pd.concat(dados_bps_20_26, ignore_index=True)

# salvar o dataframe unificado em um arquivo CSV
tabela_unificada.to_csv("BPS_20_26_MariaHelena.csv", index=False, sep=";")

print("Dataframe unificado criado com sucesso")

```
